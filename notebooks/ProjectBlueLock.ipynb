This project will seek to be a data-driven soccer player discovery and profiling system that identifies players whose performance characteristics, development trajectory, and playing profile resemble successful players or fit predefined club/role profiles.

-For example, among players in this population, this player exhibits an unusual combination of ball progression, chance creation, defensive contribution, age-adjusted production, and physical characteristics that closely resembles players who later succeeded in this role.

Ultimate Goal: Predict which young players will become elite 

Eventually build an interface where someone selects: 

Position: CM
Age: ≤ 21
League: Any
Profile: Progressive Playmaker
Reference player: Luka Modrić

and receives candidate players, percentile profiles, similarity metrics, confidence indicators, development trends, and comparison visualizations.

Player Profiling System 

Create a quantiative representation of a player 

Forward/Midfielder/Defender/Goalkeeper represented by something like: 

X=[x1,x2,x3,…,xn]   which is just a vector in n dimensional space

where the features might represent things such as progressive passes, progressive carries, expected assists, ball recoveries, pressures, interceptions, passing under pressure, shot creation, touches in different zones, and so on. Suddenly every player becomes a point in an n-dimensional feature space

Player Similarity System

Once players exist as vectors we can ask

How similar are Player A and Player B? - how do we determine this? Potentially: cosine similarity or euclidean distance

Which would then look something like 

PLAYER: Young midfielder X

98% profile similarity → Player A
94% → Player B
91% → Player C

But importantly, we'd have to explain why they're similar rather than presenting a magical score.

Talent Discovery 

Suppose a player has:

similar characteristics to high-performing players;
unusually strong metrics relative to players their own age;
strong performance despite playing in a weaker team;
statistically significant improvement over several seasons;
high performance across multiple dimensions rather than one flashy statistic.

That player could receive a candidate score.

Conceptually:

CandidateScore=f(Performance,Age,Trajectory,Similarity,LeagueStrength,RoleFit,Consistency,Minutes,Context)

This function could be learned through machine learning

Could our system have identified players who eventually became high-level professionals?

Project Charter v0.2

Problem

Soccer clubs have limited scouting resources and must evaluate large numbers of potential players across leagues, teams, positions, and competitive environments.

Primary user

Club recruitment analysts and sporting directors.

Decision

Determine which players deserve further scouting based on statistical characteristics and desired player profiles.

Initial candidate pool

Young senior-level professional outfield players, with our exact age/leagues determined after investigating available data.

Initial system output

For each player:

statistical profile,
strongest characteristics,
percentile rankings,
professional player similarities,
searchable player attributes,
cross-player comparison.
Long-term goal

Expand the system toward identifying undervalued and emerging talent, incorporating development trajectory, tactical fit, league strength, age, physical/tracking information, and eventually youth players.


Project Roadmap 

Phase 1 — Player Representation
Can we accurately describe professional players quantitatively?

Phase 2 — Player Similarity
Can we identify genuinely comparable playing profiles?

Phase 3 — Player Discovery
Can we search for players possessing specific combinations of characteristics?

Phase 4 — Contextualization
Can we account for position, team possession, league, age, minutes, tactical environment, etc.?

Phase 5 — Development & Potential
Can historical player trajectories tell us something meaningful about future outcomes?

Phase 6 — Club Fit
Can we characterize what a specific club requires and identify matching players?

Phase 7 — Youth Identification
Can we adapt the framework to less-developed players and limited youth data?

Phase 8 — Product
Can someone actually use the system effectively through SQL, Python, Tableau/Power BI and eventually an application?
