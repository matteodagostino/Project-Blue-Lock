```python
#copy in a df only the specific columns i want to currently analyze for pass mapping
keepColumns = [
    "player",
    "team",
    "position",
    "location",
    "pass_end_location",
    "pass_length",
    "pass_recipient",
    "pass_outcome",
    "play_pattern",
    "minute",
    "second"
]

#create a passing-only DataFrame
passingDF = CFCvARSvGW1.loc[CFCvARSvGW1["type"] == "Pass",keepColumns].copy()

#verify the results
print(passingDF.shape)
passingDF.head()
```

```python
#location is a string, pass end location is a string, pass length is a float64 so certain things need to be modified

#using ast library to safely convert a string representation of a list into an actual python list
import ast 

#convert coordinate strings into python lists
passingDF["location"] = passingDF["location"].apply(ast.literal_eval)
passingDF["pass_end_location"] = passingDF["pass_end_location"].apply(ast.literal_eval)

#validate
print(type(passingDF["location"].iloc[0]))
print(passingDF["location"].iloc[0])
```

```python
#separate the x and y coordinates that way we can use them later to understand more in depth passing metrics 
passingDF[["start_x", "start_y"]] = pd.DataFrame(passingDF["location"].tolist(), index=passingDF.index)
passingDF[["end_x", "end_y"]] = pd.DataFrame(passingDF["pass_end_location"].tolist(), index=passingDF.index)

#verify results 
passingDF[["player", "start_x", "start_y", "end_x", "end_y", "pass_length"]].head()
```

```python
passingDF.head(5)
```

```python
#calculate vectors using numpy
import numpy as np

#calculate the change in x and y  using x2-x1, y2-y1
passingDF["delta_x"] = passingDF["end_x"] - passingDF["start_x"]
passingDF["delta_y"] = passingDF["end_y"] - passingDF["start_y"]

#calculate the magnitude aka the length between each pass 
passingDF["calculated_length"] = np.hypot(passingDF["delta_x"], passingDF["delta_y"])

#validate 
passingDF[["player", "delta_x", "delta_y", "pass_length", "calculated_length"]].head()
```

```python
#data validation even though we can see from head that isnt enough to be sure that everything was caluclated perfectly
passingDF["length_error"] = (passingDF["calculated_length"] - passingDF["pass_length"]).abs()

#sumamrize the errors 
passingDF["length_error"].describe()
```

```python
#building first player passing map 
#the idea right now is validate the passing visualization using one player to start 
#then build on that foundation until we can display both teams and filter by player
#the goal is to eventually open the match and be able to see all the passing data on the full pitch and select a player to isolate their passes


#select the player to analyze 
player_name = "Francesc Fàbregas i Soler"

#filter the passing dataset
player_passes = passingDF.loc[passingDF["player"] == player_name].copy()

#check the results
print("Total passes:", len(player_passes)) 
player_passes[["player", "start_x", "start_y", "end_x", "end_y", "pass_outcome"]].head()

#separating the passes by outcome
completed = player_passes.loc[player_passes["pass_outcome"].isna()]
other_outcomes = player_passes.loc[player_passes["pass_outcome"].notna()]

#check results 
print("Completed:", len(completed))
print("Other outcomes:", len(other_outcomes))

#libraries to use
from mplsoccer import Pitch
import matplotlib.pyplot as plt

#create the pitch
pitch = Pitch(pitch_type = "statsbomb", pitch_color='grass', line_color='white', stripe=True)
fig, ax = pitch.draw(figsize=(14,9))

#plot completed passes
pitch.arrows(
    completed["start_x"],
    completed["start_y"],
    completed["end_x"],
    completed["end_y"],
    ax=ax,
    color="#7ee081",
    width=1.5,
    headwidth=4,
    headlength=5,
    alpha=0.7,
    label="Completed"
)

#plot other pass outcomes
pitch.arrows(
    other_outcomes["start_x"],
    other_outcomes["start_y"],
    other_outcomes["end_x"],
    other_outcomes["end_y"],
    ax=ax,
    color="#ff7474",
    width=1.5,
    headwidth=4,
    headlength=5,
    alpha=0.7,
    label="Other outcomes"
)

ax.legend(loc="upper left")

ax.set_title(
    f"{player_name} | Chelsea vs Arsenal 2015/16",
    fontsize=15,
    pad=15
)

plt.show()
```

## Project Blue Lock — Passing Map Pipeline (Phase 1)

### Objective
Develop a reusable pipeline to transform raw StatsBomb event data into individual player passing maps, allowing me to visualize passing locations, directions, distances, and outcomes on a soccer pitch.

### Data Preparation and Cleaning

Successfully created a passing-only DataFrame (`passingDF`) by copying and filtering the original match dataset to include only pass events and the specific variables relevant to the analysis. This allowed me to preserve the original dataset while working with a smaller, more manageable collection of passing data.

The `location` and `pass_end_location` variables were initially stored as strings after importing the CSV file. These were converted into Python lists so that their individual x and y coordinates could be extracted.

Four new numerical columns were created:

- `start_x`, `start_y` — Starting coordinates of each pass.
- `end_x`, `end_y` — Ending coordinates of each pass.

This transformation allows me to perform mathematical operations on the coordinates and eventually develop more advanced passing metrics, such as progressive passes, passes into the final third, passes into the penalty area, passes in defensive third and so on.

### Vector Calculations and Validation

Using NumPy, I calculated the change in horizontal and vertical position (`delta_x` and `delta_y`) between the starting and ending coordinates of each pass.

I then used the Euclidean distance formula to calculate the magnitude (length) of each passing vector.

To validate the calculations, I compared the calculated distances against StatsBomb's existing `pass_length` variable. The differences were extremely small, indicating strong numerical agreement and supporting the correctness of our coordinate transformations and distance calculations.

### Passing Map Visualization

After validating the calculations, I successfully created the first individual player passing map.

The process involved:

1. Selecting a specific player and filtering `passingDF` to create a temporary DataFrame containing only that player's passes.
2. Separating completed passes from other recorded outcomes using the `pass_outcome` variable. I identified that `NaN` generally indicates a completed pass, while populated values describe other outcomes such as incomplete passes, passes out of play, and offsides.
3. Creating a soccer pitch using the `mplsoccer` library.
4. Using Matplotlib and `mplsoccer` to plot the passing vectors from their starting coordinates to their ending coordinates.
5. Applying different colors to distinguish completed passes from other pass outcomes.

### Results and Validation

The passing map was successfully generated for Cesc Fàbregas. I then tested the same visualization code on additional players, including Santi Cazorla, Mesut Özil, and Theo Walcott.

All visualizations rendered successfully, demonstrating that the same pipeline can be reused for different players by changing the player-selection filter.

The resulting maps allow me to visually examine where players attempt passes, the direction and distance of those passes, and their recorded outcomes.
